```python
import sklearn.datasets as datasets
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn import tree
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
iris_df = pd.read_csv(r"C:\Users\User\Iris.csv")
iris_df.head()
```

```python
print(iris_df.columns.tolist())
```

```python
iris_df = iris_df.rename(columns={
    'sepal_length': 'SepalLengthCm',
    'sepal_width': 'SepalWidthCm',
    'petal_length': 'PetalLengthCm',
    'petal_width': 'PetalWidthCm',
    'species': 'Species'
})

iris_df.head()
```

```python
sns.pairplot(iris_df, hue='Species')
plt.savefig("exercise2a.png")
plt.show()
```

```python
X = iris_df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']].values
y = iris_df['Species'].values

X_train, X_test, y_train, y_test = train_test_split(
    X, y, train_size=0.9, random_state=1
)
```

```python
# scikit learn decision tree model training
clf = tree.DecisionTreeClassifier()
clf = clf.fit(X_train, y_train)
```

```python
prediction = clf.predict(X_test)
prediction
```

```python
# evaluation for multi class classification
print(classification_report(y_test, prediction))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_estimator(clf, X_test, y_test)
plt.savefig("exercise2b.png")
plt.show()
```

```python
X = iris_df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']].values

y = iris_df['Species'].values
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    train_size=0.9,
    random_state=1
)
```

```python
from sklearn import tree

clf = tree.DecisionTreeClassifier()

clf = clf.fit(X_train, y_train)
```

```python
prediction = clf.predict(X_test)

prediction
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, prediction))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_estimator(
    clf,
    X_test,
    y_test
)

plt.savefig("exercise2b.png")
plt.show()
```
